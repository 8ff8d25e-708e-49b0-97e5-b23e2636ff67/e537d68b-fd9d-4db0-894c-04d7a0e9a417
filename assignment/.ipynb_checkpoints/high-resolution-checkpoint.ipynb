```vadalog
% We're going to create the person nodes
@bind("dataDictionaryI","csv useHeaders=true","./disk/assignment/data/high-resolution/dictionaries","DynamicHormoneDiagnostics_DataDictionary_2022-03-02.csv").
@bind("formDictionary1","csv useHeaders=true, delimiter='\t'","./disk/assignment/data/high-resolution/dictionaries","ultradian-912.csv").
@bind("formDictionary2","csv useHeaders=true, delimiter='\t'","./disk/assignment/data/high-resolution/dictionaries","ultradian-914.csv").
@bind("formControls1I","csv useHeaders=true","./disk/assignment/data/high-resolution","form_912_controls.csv").
@bind("formControls2I","csv useHeaders=true","./disk/assignment/data/high-resolution","form_914_controls.csv").
@bind("hrTimeSeriesI","csv useHeaders=true","./disk/assignment/data/high-resolution","md_data_controls.csv").
@bind("hrMetaI","csv useHeaders=true","./disk/assignment/data/high-resolution","redcap_controls.csv").

@model("dataDictionary", "['MasterID:string', 'FormName:string', 'SectionHeader:string', 'FieldType:string','FieldLabel:string','ChoicesCalculationsORSliderLabels:string','FieldNote:string','TextValidationTypeORShowSliderNumber:string']").
dataDictionary(A,B,C,D,E,F,G,H) :-
    dataDictionaryI(A,B,C,D,E,F,G,H,_,_,_,_,_,_,_,_,_,_,_).

@model("formDictionary", "['FormId:int','ItemId:int','VarName:string','LoincScale:string','UnitStr:string','ItemText:string','Expression:string','MinExpression:double','MaxExpression:double','ItemHelp:string','Decimals:int','OrderNumber2:int','OptionText:string']").
formDictionary(A,B,C,D,E,F,G,H,I,J,K,L,M) :-
    formDictionary1(A,B,C,D,E,F,G,H,I,J,K,L,M).
formDictionary(A,B,C,D,E,F,G,H,I,J,K,L,M) :-
    formDictionary2(A,B,C,D,E,F,G,H,I,J,K,L,M).

formControls1(MasterID,GenderId,ULT_SMOKER,HEIGHT) :-
    formControls1I(MasterID,GenderId,GroupName,ULT_SMOKER,HEIGHT,CUSH_CAUSE,PETROSAL_SAMPLING_POSITI,CRH_POSITIVE,HIGH_DOSE_DEX_POSITIVE,AVS_OUTCOME,CORT_BLOOD_MORNING_DX,ACTH_MORNING_DX,CORT_BLOOD_NIGHT_DX,CORT_SALIVA_DX,CORT_24H_URINE_DX,SHORT_DEX_CORT_VALUE_DX,ULT_ALDOSTER_DX,ULT_ALDOSTER_ATHENS_DX,ALDO_RRR_DX,ULT_RENIN_BERGEN_DX,ULT_RENIN_ATHENS_DX,ULT_RENIN_BRISTOL_DX,RENIN_RRR_DX,ULT_ARR_DX,ARR_RRR_DX,K_LOWEST,K_LLN,ALDO_DX_CONF,ALDO_LOWEST_CONF,ALDO_LOWEST_CONF_ATHENS,ALDO_IMAGING,ALDO_AVS,ALDO_SUBTYPE,ALDO_BMP_MED_MAX,COMORBID_T2DM,COMORBID_HT,COMORBID_CARDIO,COMORBID_CEREBRO,COMORBID_OSTEO,COMORBID_THROMBO,COMORBID_OTHER,COMORBID_OTHER_DESC).

personMeasurements(MasterID,WAIST,HIP,WAIST_HIP_RATIO,SYSBP,DIABP,WEIGHT,BMI,Age) :-
    formControls2I(MasterID,GenderId,GroupName,HEALTHY_SUBSTUDY,SAMPLING_SUCCESS,SAMPLING_APPROVAL_STATUS,SAMPLING_COMMENT,SAMPLING_ESTROGEN,WAIST,HIP,WAIST_HIP_RATIO,SYSBP,DIABP,WEIGHT,BMI,SAMPLING_START_HOUR,SAMPLING_START_MIN,SAMPLING_CATHETER,SAMPLING_SAMPLER_ID,SAMPLING_STOP_HOUR,SAMPLING_STOP_MIN,SAMPLE_LAST_WELL,SAMPLES_DISCARDED,SAMPLE_OUTLIER,SAMPLE_OUTLIER_TEXT,SAMPLING_RXN,SAMPLING_QUALITY,SALIVA_1_HOUR,SALIVA_1_MIN,SALIVA_2_HOUR,SALIVA_2_MIN,SALIVA_3_HOUR,SALIVA_3_MIN,CUSH_TIMELINE,CUSH_HIST_DX,CUSH_SURG_DATE,CUSH_SURG_MONTHS_AGO,CUSH_POSTOP_CORT,CUSH_POSTOP_INHIBITOR,CORT_BLOOD_MORNING,ACTH_MORNING,CUSH_SYNACTH_CORT,CORT_24H_URINE,CORT_BLOOD_NIGHT,CORT_SALIVA,SHORT_DEX_CORT_VALUE,CUSH_REMISSION,ALDO_CA_BLOCKER,ALDO_ALPHA_BLOCKER,ALDO_MOXONIDE,ALDO_K_SUPP,ALDO_K_VALUE,ALDO_ESTROGEN,ALDO_TIMELINE,ALDO_HIST_DX,ALDO_SURG_DATE,ALDO_BP_MED_POSTOP,ULT_ALDOSTERONE,ULT_ALDOSTERONE_ATHENS,ALDO_RRR,ULT_RENIN_BERGEN,ULT_RENIN_ATHENS,ULT_RENIN_BRISTOL,RENIN_RRR,ULT_ARR,ARR_RRR,PAI_DX_DATE,PAI_HYPO_THY,PAI_HYPER_THY,PAI_T1DM,PAI_POI,PAI_CELIAC,PAI_AAG,GC_REPLACEMENT_TYPE,GC_PUMP_DOSE,GC_DOSE_1_TIME,GC_ACTUAL_DOSE_1,HC_EQUIVALENT_DOSE_1,GC_DOSE_2_TIME,GC_ACTUAL_DOSE_2,HC_EQUIVALENT_DOSE_2,GC_DOSE_3_TIME,GC_ACTUAL_DOSE_3,HC_EQUIVALENT_DOSE_3,GC_DOSE_4_TIME,GC_ACTUAL_DOSE_4,HC_EQUIVALENT_DOSE_4,GC_DOSE_5_TIME,GC_ACTUAL_DOSE_5,HC_EQUIVALENT_DOSE_5,GC_DOSE_6_TIME,GC_ACTUAL_DOSE_6,HC_EQUIVALENT_DOSE_6,GC_DOSE_7_TIME,GC_ACTUAL_DOSE_7,HC_EQUIVALENT_DOSE_7,MR_DOSE_1_TIME,MR_DOSE_1_MG,MR_DOSE_2_TIME,MR_DOSE_2_MG,MR_DOSE_3_TIME,MR_DOSE_3_MG,Age).

@model("gender","['gender_id(ID):string']").
gender("Male").
gender("Female").


@model("person", "['person_id(ID):int','gender:gender','smoker:boolean','height:double','waist:double','hip:double','waist_hip_ratio:double','sysBP:int','diaBP:int','weight:double','BMI:double','age:int']").
person(ID,G,S,HEIGHT,WAIST,HIP,WAIST_HIP_RATIO,SYSBP,DIABP,WEIGHT,BMI,Age) :- 
    formControls1(ID,G,S,HEIGHT),
    personMeasurements(ID,WAIST,HIP,WAIST_HIP_RATIO,SYSBP,DIABP,WEIGHT,BMI,Age).

@bind("person","csv useHeaders=true","./disk/assignment/output/csvs","person.csv").

@output("person").
```
