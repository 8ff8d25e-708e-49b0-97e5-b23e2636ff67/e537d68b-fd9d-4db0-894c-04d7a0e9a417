```vadalog
@bind("hormoneI","csv useHeaders=true","./disk/assignment/data/csvs","hormones-detailed.csv").
@bind("bindingI","csv useHeaders=true","./disk/assignment/data/csvs","hormone-bindings-b.csv").
@bind("receptorI","csv useHeaders=true","./disk/assignment/data/csvs","receptors.csv").

hormone(HormoneId,HormoneGroupId,Abbrev,Name,ReceptorSwissProt,Receptor,Modulator,Description) :-
    hormoneI(HormoneId,HormoneGroupId,Abbrev,Name,ReceptorSwissProt,Receptor,Modulator,Description).

receptor(Type,FamilyId,FamilyName,TargetId,TargetName,SubunitId,SubunitName,TargetSystematicName,TargetAbbrev,Synonyms,HGNCId,HGNCSymbol,HGNCName,HumanGeneticLocalisation,HumanNucleotideRefSeq,HumanProteinRefSeq,HumanSwissProt,HumanEntrezGene,HumanEnsemblGene) :-
    receptorI(Type,FamilyId,FamilyName,TargetId,TargetName,SubunitId,SubunitName,TargetSystematicName,TargetAbbrev,Synonyms,HGNCId,HGNCSymbol,HGNCName,HumanGeneticLocalisation,HumanNucleotideRefSeq,HumanProteinRefSeq,HumanSwissProt,HumanEntrezGene,HumanEnsemblGene).


targets(HormoneId,ReceptorId,ReceptorSwissProt,Affinity,Type) :-
    hormone(HormoneId,_,_,_,ReceptorSwissProt,ReceptorId,Affinity,Description),
    receptor(_,_,_,_,_,_,_,_,_,_,_,ReceptorId,_,_,_,_,ReceptorSwissProt,_,_).

targets(HormoneId,ReceptorId,ReceptorSwissProt,Affinity,Type) :-
    hormone(HormoneId,_,_,_,_,_,_,_),
    bindingI(HormoneId,ReceptorId,ReceptorSwissProt,Affinity,Type),
    receptor(_,_,_,_,_,_,_,_,_,_,_,ReceptorId,_,_,_,_,ReceptorSwissProt,_,_).

hormoneTargets(Hormone,Receptor,Affinity,Type,Description) :-
    hormone(HormoneId,_,_,Hormone,_,_,_,Description),
    targets(HormoneId,ReceptorId,ReceptorSwissProt,Affinity,Type),
    receptor(_,_,_,_,Receptor,_,_,_,_,_,_,ReceptorId,_,_,_,_,ReceptorSwissProt,_,_).

receptorType(Type) :- receptor(Type,_,_,_,_,_,_,_,_,_,_,_,_,_,_,_,_,_,_).
receptorFamily(FamilyId,FamilyName) :- receptor(_,FamilyId,FamilyName,_,_,_,_,_,_,_,_,_,_,_,_,_,_,_,_).

partOf(FamilyId,Type) :- 
    receptorFamily(FamilyId,_),
    receptorType(Type),
    receptorI(Type,FamilyId,_,_,_,_,_,_,_,_,_,_,_,_,_,_,_,_,_).
partOf(TargetId,FamilyId) :- 
    receptor(_,FamilyId,_,TargetId,_,_,_,_,_,_,_,_,_,_,_,_,_,_,_),
    receptorFamily(FamilyId,_).

partOfT(FamilyId,Type) :- partOf(FamilyId,Type).
partOfT(A,B) :- partOfT(FamilyId,Type)

@output("hormoneTargets").

```
